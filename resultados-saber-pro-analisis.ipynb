## Depuracion del dataset

```python
# Verificar si hay valores nulos en el DataFrame
valores_nulos = df.isnull().sum()
print("Valores nulos por columna:")
print(valores_nulos)
```

```python
# Eliminar filas con valores nulos en las columnas de ESTU_DEPTO_RESIDE
# Se eliminan a estos usuarios ya que es imposible determinar su ubicación geográfica y esto puede afectar el análisis de los datos.
df = df.dropna(subset=['ESTU_DEPTO_RESIDE'])

# Verificar el número de filas restantes
valores_nulos = df.isnull().sum()
print("Valores nulos por columna:")
print(valores_nulos)
```

```python
# Verificar si hay filas duplicadas en la columna ESTU_CONSECUTIVO
duplicados = df.duplicated(subset=['ESTU_CONSECUTIVO'], keep=False)
print("Filas duplicadas en la columna ESTU_CONSECUTIVO:")
print(df[duplicados])
```

```python
# Eliminar filas duplicadas basadas en la columna ESTU_CONSECUTIVO, manteniendo la primera ocurrencia
df = df.drop_duplicates(subset=['ESTU_CONSECUTIVO'], keep='first')

# Verificar si hay filas duplicadas después de la eliminación
duplicados = df.duplicated(subset=['ESTU_CONSECUTIVO'], keep=False)
print("Filas duplicadas en la columna ESTU_CONSECUTIVO después de la eliminación:")
print(df[duplicados])
```

```python
# Verificar si hay columnas de tipo float que deberían ser enteros, en caso tal de presente valores con decimales
# se debe aplicar piso o techo
columnas_float = df.select_dtypes(include=["float", "float64"]).columns

for columna in columnas_float:
    # Ignorar valores nulos
    valores = df[columna].dropna()

    # Detectar valores que tienen parte decimal
    tiene_decimal = ~np.isclose(valores % 1, 0)

    print(f"\nColumna: {columna}")
    print(f"Total de valores: {len(valores)}")
    print(f"Valores con decimales: {tiene_decimal.sum()}")

    if tiene_decimal.any():
        print("Ejemplos:")
        print(valores[tiene_decimal].head())
    else:
        print("Todos los valores son enteros aunque la columna sea float.")
```

```python
# Convertir las columnas float sin perder los valores faltantes.
# Int64 es el tipo entero nullable de pandas y admite valores NA.
columnas_float = df.select_dtypes(include=["float", "float64"]).columns

for columna in columnas_float:
    valores_no_finitos = ~np.isfinite(df[columna])
    cantidad_no_finitos = valores_no_finitos.sum()

    df[columna] = (
        df[columna]
        .replace([np.inf, -np.inf], np.nan)
        .astype("Int64")
    )

    print(f"Columna {columna} convertida a Int64; valores no finitos conservados como NA: {cantidad_no_finitos}")
```
